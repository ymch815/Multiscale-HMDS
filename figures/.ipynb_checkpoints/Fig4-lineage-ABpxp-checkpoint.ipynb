```python
def compute_apindex_assign(name,ords):
    if len(name)<8:
        return np.nan
    name = name[8:]
    if name in ords:
        return ords[name]
    else:
        return np.nan
```

```python
# load data: individual embedding, in 3d, 5d, 7d
pcoords = np.loadtxt('../MultiscalehMDS_feature/result/FeatMat_Celegans_PCA50_lineage_ABpxp/cls_50_Nn_20_Nd_7_min_3_max_400_corr_0_coords_7562.txt')
inds = np.loadtxt('../MultiscalehMDS_feature/result/FeatMat_Celegans_PCA50_lineage_ABpxp/cls_50_Nn_20_Nd_7_min_3_max_400_corr_0_inds_after_prep.txt')

pcoords = pcoords[np.argsort(inds)]
```

```python
metadata = pd.read_csv('../data/metadata/Celegans_PCA50_lineage_ABpxp_metadata.csv',index_col=0)
```

```python
# compute features on lineage
df_toplot = metadata.reset_index()
# label lineage
generation_label = np.array([len(name)-7 for name in metadata.lineage])
# label a-p

ords = {'pp':3.0,'pa':2.0,'ap':1.0,'aa':0.0}

ords_shuffle = {'pp':3.0,'pa':1.0,'ap':2.0,'aa':0.0}
ap_label = np.array([compute_apindex_assign(name,ords) for name in metadata.lineage])
ap_label_shuffled = np.array([compute_apindex_assign(name,ords_shuffle) for name in metadata.lineage])

df_toplot['generation_label'] = generation_label
df_toplot['ap_label'] = ap_label
df_toplot['ap_label_shuffled'] = ap_label_shuffled
```

```python
N_sample = pcoords.shape[0]
data_recenter = emb.perform_recenter(pcoords,np.arange(N_sample))
```

```python
native3d = emb.to_native(data_recenter)
radii = np.linalg.norm(native3d,axis=1)
```

```python
# visualize two sister lineages: ABpxpppp and ABpxpppa
print(np.unique(df_toplot[df_toplot.lineage.str.startswith('ABpxppaa',na=False)].lineage))
print(np.unique(df_toplot[df_toplot.lineage.str.startswith('ABpxppap',na=False)].lineage))
```

```python
'''
fit 1-d geodesic in high-dimensional space
along division direction / along a-p direction
how: along division direction: label the sample with generations, find direction maximizing that change. 
along a-p direction: label the sample with a-p preference, find direction maximizing that change

n-d hyperbolic pcoords -> map to tangent space -> choose direction -> projection -> find direction maximize the change
visualize projections
'''

# tangent space
tangent_coords = emb.logmap(data_recenter)
```

```python
def mygoal(proj_target,coords,metric):
    proj_target = proj_target/np.linalg.norm(proj_target)
    projs = np.dot(coords, proj_target)  
    corr,p = stats.pearsonr(projs,metric)
    return -corr
```

```python
def proj_gen_ap(lineage_name,df_toplot,tangent_coords):
    curr = lineage_name
    sel = np.where((df_toplot.lineage.str.startswith(curr,na=False)))[0]
    sel2 = np.where((df_toplot.lineage.str.startswith(curr,na=False))&(df_toplot.generation_label==3))[0]
    print(np.unique(df_toplot['lineage'][sel2]))
    x0 = np.mean(tangent_coords[np.isin(df_toplot.lineage.astype(str),[curr+'p',curr+'a']),:],
                          axis=0)-np.mean(tangent_coords[df_toplot.lineage.astype(str)==curr,:],axis=0)
    res1 = minimize(mygoal, x0, args=(tangent_coords[sel,:],df_toplot['generation_label'][sel]), method='L-BFGS-B',
                   tol=1e-12)
    res2 = minimize(mygoal, x0, args=(tangent_coords[sel2,:],df_toplot['ap_label'][sel2]), method='L-BFGS-B',
                   tol=1e-12)
    res3 = minimize(mygoal, x0, args=(tangent_coords[sel2,:],df_toplot['ap_label_shuffled'][sel2]), method='L-BFGS-B',
                   tol=1e-12)
    df_return = df_toplot.copy()
    df_return['gen_proj'] = np.dot(tangent_coords,res1.x)
    df_return['ap_proj'] = np.dot(tangent_coords,res2.x)
    df_return['ap_proj_shuffled'] = np.dot(tangent_coords,res3.x)
    return df_return
```

```python
dfs = []
lilist = ['ABpxpppp','ABpxpppa','ABpxppap','ABpxppaa','ABpxpaap','ABpxpaaa']
for li_ in lilist:
    dfs.append(proj_gen_ap(li_,df_toplot,tangent_coords))
```

```python
def set_axs(axs,x='',y='',title=''):
    axs.set_title(title,fontsize=15)
    axs.set_xlabel(x,fontsize=15)
    axs.set_ylabel(y,fontsize=15)
    axs.tick_params(axis="x", labelsize=12)  
    axs.tick_params(axis="y", labelsize=12)
```

```python
# plot 6 different lineages
fig,axs = plt.subplots(3,2,figsize=(10,10))
axs = axs.flatten()
for i in range(6):
    sn.scatterplot(data=dfs[i][(dfs[i].lineage.str.startswith(lilist[i]))&(dfs[i]['generation_label']<=3)],
               x='gen_proj',y='ap_proj',hue='lineage',palette=colormaps[lilist[i]],ax=axs[i],s=40,edgecolor="none")
    axs[i].get_legend().remove()
    set_axs(axs[i],'Generation Axis Projection','A-P Axis Projection',
           lilist[i])
    
plt.tight_layout()
```

```python
# silhouette_score
from sklearn.metrics import silhouette_score
```

```python
sscore = []
for i in range(6):
    df_ = dfs[i][(dfs[i].lineage.str.startswith(lilist[i]))&(dfs[i]['generation_label']==3)]
#     print(np.unique(df_['lineage']))
    sscore.append(silhouette_score(np.array(df_[['gen_proj','ap_proj']]),df_['lineage']))
sscore_pca = []
for i in range(6):
    df_ = dfs_pca[i][(dfs_pca[i].lineage.str.startswith(lilist[i]))&(dfs[i]['generation_label']==3)]
#     print(np.unique(df_['lineage']))
    sscore_pca.append(silhouette_score(np.array(df_[['gen_proj','ap_proj']]),df_['lineage']))
df_ = pd.DataFrame({'scores':sscore+sscore_pca,'lineage':lilist+lilist,'Method':['Multiscale H-MDS']*6+['PCA']*6})
fig,ax = plt.subplots(figsize=(6,4))
sn.barplot(data=df_,x='lineage',hue='Method',y='scores',ax=ax)
set_axs(ax,'Lineage','Silhouette Score',)
ax.legend(fontsize=12)
ax.tick_params(axis="x", labelsize=12,rotation=20)  
plt.tight_layout()
```
